We consider $f(x) = x^3 - 2$

## Simple Newton
1. Plot the graph of $f(x)$. Identify the root graphically, and numerically.

```python
def f(x):
    return x**3 - 2

x_sol = 2**(1/3)

import matplotlib.pyplot as plt
from numpy import linspace
X = linspace(0,2,100)
plt.vlines(x_sol, -2, 0)
plt.plot(X, f(X))
plt.grid()
plt.show()
```

2. Compute the derivative of $f$

```python
def dfdx(x):
    return 3*x**2
```

3. Fill the following template code of a simple Newton algorithm without line seach.

```python
import numpy as np

def newton_simple(f, df, x0, maxit=100, tol=1e-8):
    """
    Simple Newton method for scalar root-finding.
    """
    x = [x0]
    for k in range(maxit):
        fx = f(x[-1])
        dfx = df(x[-1])
        descent = - fx / dfx
        x.append(x[-1] +  descent)
        
        if abs(fx*descent)**0.5 < tol: # stop criterion
            print(f"Newton is converged.")
            return np.array(x)
    print("Newton has failed.")
    return np.array(x)

```

4. Apply the Newton method to solve $f(x) = 0$. Change the starting point. Does the method always converge? Comment.

```python
x0 = 1e-100
x_hist = newton_simple(f, dfdx, x0)
plt.plot(x_hist, 'o')
plt.show()
```

## Line search

5. Fill the following template to add a line search to the Newton algorithm.

```python
def linesearch_test(f, x, alpha, i, descent, s):
    F = lambda x : f(x)**2
    try:
        test = F(x + alpha**i * descent) >= ( 1 - 2*s*alpha**i ) * F(x)
        return test
    except:
        return True
        
    
def newton_linesearch(f, df, x0, alpha = np.array(0.5), s = 0, maxit_linesearch = 1000, maxit=100, tol=1e-8):
    """
    Line-search Newton method for scalar root-finding.
    """
    x = [x0]
    for k in range(maxit):
        fx = f(x[-1])
        dfx = df(x[-1])
        descent = - fx / dfx
        
        i = 0
        while linesearch_test(f, x[-1], alpha, i, descent, s):
            if i>maxit_linesearch:
                print("Linesearch has failed")
                return np.array(x)
            i += 1
        
        increment = alpha**i * descent
        x.append(x[-1] +  increment)
        
        if abs(fx*increment)**0.5 < tol: # stop criterion
            print(f"Newton is converged.")
            return np.array(x)
    print("Newton has failed.")
    return np.array(x)
```

6. Apply the Newton method with line-search to solve $f(x) = 0$. Change the starting point. Does the method always converge? Comment.

```python
x0 = 1e-100
x_hist = newton_linesearch(f, dfdx, x0)
plt.hlines(x_sol, 0, len(x_hist))
plt.plot(x_hist, 'o')
plt.show()
```

7. Is the q-convergence rate quadratic?

```python
def plot_convergence(values, q = 1):
    """ Plot q-convergence to 0 of a list of values """
    ax1 = plt.semilogy(np.abs(values), "bo") 
    plt.ylabel("$|x_n - x_{ref}|$", color = 'b')
    plt.xlabel('Itérations')
    ax2 = plt.gca().twinx()
    mu = np.abs(values[1:]) / np.abs(values[:-1])**q
    ax2.plot(mu, 'ro')
    plt.ylabel(f'$|x_{{n+1}} - x_{{ref}}| / |x_{{n}} - x_{{ref}}|^{q}$', color = 'r')
    plt.grid(color = [1,0.8,0.8])
    plt.title(f"Convergence of the algorithm\n ${q = }$ ; $\\mu \\simeq {mu[-1] :.5f}$ ")
    
plot_convergence(x_hist - x_sol, q = 2)
```
