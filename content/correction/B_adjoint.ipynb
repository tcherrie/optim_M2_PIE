## B) Adjoint method

Consider the following state equation:

$$ \begin{bmatrix}
1 & 2x & x^2 \\
2x & 2 & x^3 \\
x^2 & x^3 & 1\\
\end{bmatrix}
\begin{bmatrix}
u_0 \\ u_1 \\ u_2
\end{bmatrix} - \begin{bmatrix}
1 \\ 2 \\ 3
\end{bmatrix} = 0$$

So that the state $u$ depends implictely on the design variable $x$. 
We are interested in minimizing the following criterion:

$C(u,x) = 3x + |u|^2 $

Current design variable is $x_0=1$.

1. Compute $u(x_0)$ and $C(u_0, x_0)$.

```python
from scipy.linalg import solve
from numpy import array

def K(x):
    return array([[1, 2*x, x**2], [2*x, 2, x**3], [x**2, x**3, 1]])

rhs = array([[1],[2],[3]])

x0 = 1
u0 = solve(K(x0), rhs)
print(u0)

from numpy.linalg import norm
def C(u,x):
    return 3*x + norm(u)**2

C0 = C(u0, x0)
print(f"{C0 = :.2f}")
    
```

2. Compute $\partial_u S(x_0)$

```python
def partial_S_partial_u(x):
    return K(x)

print(partial_S_partial_u(x0))
```

3. Formulate and solve the adjoint problem.

```python
def dC_du(u):
    return 2 * u

p0 = solve(partial_S_partial_u(x0).T, - dC_du(u0))

print(p0)
```

4. Compute the total derivative of $C$ with respect to $x$

```python
partial_C_partial_x = 3

def partial_S_partial_x(x, u):
    u = u[:,0]
    return array([[2*u[1] + 2*x*u[2] ], \
                  [2*u[0] + 3*x**2*u[2]],\
                  [2*x*u[0] + 3*x**2*u[1]]])

def dC_dx(u, p, x):
    return partial_C_partial_x + (p.T @ partial_S_partial_x(x,u)).squeeze()

print(dC_dx(u0, p0, x0))
```

5. Check your result with finite differences.

```python
h = 1e-6
dC_dx_FD = (C(solve(K(x0+h), rhs), x0+h) - C0) / h
print(dC_dx_FD)
```
